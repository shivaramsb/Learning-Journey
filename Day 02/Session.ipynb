**Relational Operators**

```python
#Comparison operators are used to compare two values
```

```python
print(4>5)
print(4<5)
print(4>=5)
print(4<=5)
print(4==4)
print(4!=4)
```

**Logical Operator**

```python
#Logical operators are used to combine conditional statements
```

- AND : Returns True if both statements are true
- OR : Returns True if one of the statements is true
- NOT : Reverse the result, returns False if the result is true

```python
print(1 and 0)
print(1 or 0)
print(not 1)
```

**Bitwise**

```python
# special operators that workes on binary operators 
```

```python
#bitwise and 
print(2 & 3) #Sets each bit to 1 if both bits are 1

#bitwise or 
print(2 | 3) #Sets each bit to 1 if one of two bits is 1

# bitwise xor : Sets each bit to 1 if only one of two bits is 1
print(2 ^ 3)
```
