## 1. Introduction to Python

### Why Python for AI?
Python is the dominant language in AI/ML because of its **simplicity** and **rich ecosystem** (libraries like NumPy, Pandas, PyTorch, TensorFlow).

### Comparison with Other Languages
| Feature | Python | C++ | Java |
| :--- | :--- | :--- | :--- |

| **Syntax** | Simple, English-like | Complex, Verbose | Verbose |
| **AI Usage** | Prototyping, Training, Serving | High-performance Inference, Robotics | Enterprise Systems |

## 3. Variables & Mutability

### Dynamic Typing
In Python, you don't declare types. The variable takes the type of the value assigned to it.

### Mutability
- **Immutable**: Object cannot be changed after creation. (e.g., `int`, `float`, `bool`, `str`, `tuple`).
- **Mutable**: Object can be changed. (e.g., `list`, `dict`, `set`).

### 💡 Interview Question: Memory Management
**Q: What happens if I do `a = 10` and then `a = 20`? Does the number 10 change?**  
**A:** No. Integers are immutable. 
1. `10` is created in memory.
2. `a` points to `10`.
3. When you say `a = 20`, a new object `20` is created.
4. `a` now points to `20`.
5. The object `10` is eventually garbage collected if nothing else points to it.

````python
# Demonstration of Immutability
x = 10
print(f"Initial Memory Address of x: {id(x)}")

x = x + 1
print(f"New Memory Address of x: {id(x)}") 
# Notice that the address changes! '10' was not modified; '11' was created new.
````

## 4. Operators

Operators are the tools we use to manipulate data.

### Arithmetic Operators
- `+`, `-`, `*`, `/` (Standard)
- `//` (Floor Division): Rounds down to nearest integer. Useful for batching.
- `%` (Modulus): Remainder. Useful for finding even/odd numbers.
- `**` (Exponentiation): Power. Useful for loss functions (e.g., Squared Error).

### Comparison & Logical Operators
- `==`, `!=`, `>`, `<`, `>=`
- `and`, `or`, `not`

````python
# 🏭 Industry Scenario: Batching Data for Model Training
# You have 1050 images and you want a batch size of 100.

total_images = 1050
batch_size = 100

num_batches = total_images // batch_size
remaining_images = total_images % batch_size

print(f"Full Batches: {num_batches}")
print(f"Images left over: {remaining_images}")
````

## 5. Conditional Statements (`if-elif-else`)

AI logic is often just complex decision trees. `if-else` is the foundation.

### Syntax
```python
if condition:
    # do something
elif another_condition:
    # do something else
else:
    # fallback
```

````python
# 🏭 Industry Scenario: Rule-Based Chatbot Intent Classifier
# Before we have a machine learning model, we often start with rules.

user_message = "I want to return my order"

# Simple Keyword Spotting
if "return" in user_message or "refund" in user_message:
    intent = "return_request"
    print("Bot: I can help you with your return. What is your order ID?")

elif "shipping" in user_message or "track" in user_message:
    intent = "tracking_request"
    print("Bot: Please provide your tracking number.")

else:
    intent = "unknown"
    print("Bot: I didn't understand that. Can you rephrase?")
````
