```python
import boto3, json
from botocore.config import Config

session = boto3.Session(profile_name="default", region_name="us-east-1")
client = session.client("bedrock-runtime", config=Config(signature_version="v4"))

model_id = "arn:aws:bedrock:us-east-1:567886497519:inference-profile/global.amazon.nova-2-lite-v1:0"

resp = client.converse(
    modelId=model_id,
    messages=[{"role": "user", "content": [{"text": "Explain quantum computing in simple terms."}]}],
    inferenceConfig={"maxTokens": 200, "temperature": 1.0, "stopSequences": []},
)

print(resp["output"]["message"]["content"][0]["text"])
```
