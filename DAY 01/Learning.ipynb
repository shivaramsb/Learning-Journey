**Keywords & Identifiers**

compile : covert code into machine code 

Compiler : meaning or this is a program that converts high-level language code into machine code. c, c++, java are examples of high-level languages.

Interpreted : meaning or this is a program that converts high-level language code into machine code line by line. python is an example of high-level languages.

Keywords : reserved words in a programming language that have specific meanings and cannot be used as identifiers.


```python
#count of keywords in python
import keyword
print(keyword.kwlist)
print(len(keyword.kwlist))
```

```python
# identifiers: name of variables, functions, classes, modules, packages 
# keywords: reserved words in python that have specific meanings and cannot be used as identifiers

# Rules for identifiers:
# 1. identifiers can only contain letters, digits, and underscores
# 2. identifiers must start with a letter or underscore
# 3. identifiers are case-sensitive 
# 4. identifiers cannot be reserved keywords    
# 5. identifiers are case-sensitive 
# 6. special characters are not allowed only _ and . are allowed
```
